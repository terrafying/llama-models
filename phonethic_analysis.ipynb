# Phonetic Analysis Examples

This notebook demonstrates the phonetic analysis capabilities of our system, including:
- Phoneme detection and extraction
- Time-linked phonetic retrieval
- Phoneme density analysis
- Transition analysis
- Sequence analysis

```python
import sys
sys.path.append('.')

from phonetic_compositor import SyllableCompositor
from phonetic_index import PhoneticIndex, PhoneticSegment
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display
import librosa
import librosa.display
```

## Initialize the System

```python
# Initialize the compositor
compositor = SyllableCompositor()

# Example video path
video_path = "downloads/example.mp4"
```

## Basic Phoneme Analysis

```python
# Analyze the video
analysis = compositor.analyze_syllable_distribution(video_path)

print(f"Total 'e' syllables found: {analysis['total_syllables']}")
print(f"Total duration of 'e' sounds: {analysis['total_duration']:.2f} seconds")
print(f"Average syllable duration: {analysis['average_duration']:.3f} seconds")
```

## Phoneme Density Analysis

```python
# Get phoneme density over time
density = analysis['phonetic_analysis']['density']

# Plot density
times, densities = zip(*density)
plt.figure(figsize=(12, 4))
plt.plot(times, densities)
plt.title('Phoneme Density Over Time')
plt.xlabel('Time (seconds)')
plt.ylabel('Density (phonemes/second)')
plt.grid(True)
plt.show()
```

## Phoneme Transition Analysis

```python
# Get phoneme transitions
transitions = analysis['phonetic_analysis']['transitions']

# Analyze transition patterns
transition_counts = {}
for next_phoneme, _ in transitions:
    transition_counts[next_phoneme] = transition_counts.get(next_phoneme, 0) + 1

# Plot transition distribution
plt.figure(figsize=(10, 6))
plt.bar(transition_counts.keys(), transition_counts.values())
plt.title('Phoneme Transition Distribution')
plt.xlabel('Next Phoneme')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Sequence Analysis

```python
# Get phoneme sequence
sequence = analysis['phonetic_analysis']['sequence']

# Print first 10 phonemes in sequence
print("First 10 phonemes in sequence:")
for phoneme, start, end in sequence[:10]:
    print(f"{phoneme}: {start:.2f}s - {end:.2f}s")
```

## Audio Visualization

```python
# Load audio
y, sr = librosa.load(video_path)

# Plot waveform
plt.figure(figsize=(12, 4))
librosa.display.waveshow(y, sr=sr)
plt.title('Audio Waveform')
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude')
plt.show()

# Plot spectrogram
D = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)
plt.figure(figsize=(12, 4))
librosa.display.specshow(D, sr=sr, x_axis='time', y_axis='log')
plt.colorbar(format='%+2.0f dB')
plt.title('Spectrogram')
plt.show()
```

## Save Analysis Results

```python
import json

# Save analysis results
with open('phonetic_analysis_results.json', 'w') as f:
    json.dump({
        'total_syllables': analysis['total_syllables'],
        'total_duration': analysis['total_duration'],
        'average_duration': analysis['average_duration'],
        'density': density,
        'transitions': transitions,
        'sequence': sequence
    }, f, indent=2)

print("Analysis results saved to 'phonetic_analysis_results.json'")
```

## Example Incantations

Here are some example incantations for different phonetic patterns:

```python
def analyze_incantation(text: str, start_time: float, end_time: float):
    """Analyze a specific incantation in the video."""
    # Get segments in time range
    segments = compositor.analyzer.phonetic_index.get_segments_in_time_range(start_time, end_time)
    
    # Get phoneme sequence
    sequence = compositor.analyzer.phonetic_index.get_phoneme_sequence(start_time, end_time)
    
    print(f"\nAnalyzing incantation: {text}")
    print(f"Time range: {start_time:.2f}s - {end_time:.2f}s")
    print(f"\nPhoneme sequence:")
    for phoneme, start, end in sequence:
        print(f"{phoneme}: {start:.2f}s - {end:.2f}s")
    
    # Get density
    density = compositor.analyzer.phonetic_index.get_phoneme_density('e', window_size=0.5)
    if density:
        print(f"\nPhoneme density:")
        for time, d in density[:5]:
            print(f"Time {time:.2f}s: {d:.2f}")
    
    # Get transitions
    transitions = compositor.analyzer.phonetic_index.get_phoneme_transitions('e', window_size=0.1)
    if transitions:
        print(f"\nPhoneme transitions:")
        for next_phoneme, time in transitions[:5]:
            print(f"Time {time:.2f}s: e -> {next_phoneme}")
```

```python
# Example incantations
incantations = [
    ("The quick brown fox jumps over the lazy dog", 0, 5),
    ("How much wood would a woodchuck chuck", 5, 10),
    ("She sells seashells by the seashore", 10, 15)
]

for text, start, end in incantations:
    analyze_incantation(text, start, end)
```

## Save Incantations

```python
# Save incantations to file
with open('incantations.json', 'w') as f:
    json.dump(incantations, f, indent=2)

print("Incantations saved to 'incantations.json'")
```
